```python
import pandas as pd
data=pd.read_csv("HIV_Drug_Resistance_Dataset.csv")
print(data)
```

```python
data.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
b=LabelEncoder()
data["Patient_ID"]=b.fit_transform(data["Patient_ID"])
data["Subtype"]=b.fit_transform(data["Subtype"])
data["Gene"]=b.fit_transform(data["Gene"])
data["Mutation"]=b.fit_transform(data["Mutation"])
data["Drug_Class"]=b.fit_transform(data["Drug_Class"])
data["Drug_Name"]=b.fit_transform(data["Drug_Name"])
data["Resistance_Level"]=b.fit_transform(data["Resistance_Level"])
data["Treatment_Status"]=b.fit_transform(data["Treatment_Status"])
print(data)
```

```python
x=data.drop("Treatment_Status",axis=1)
y=data["Treatment_Status"]
print(x)
```

```python
print(y)
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2)
```

```python
print(xtrain)
```

```python
print(xtest)
```

```python
print(ytrain)
```

```python
print(ytest)
```

```python
from sklearn.linear_model import LogisticRegression
b=LogisticRegression()
b.fit(xtrain,ytrain)
ypred=b.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(ytest,ypred))
```

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Initialize XGBoost Classifier
xgb_model = xgb.XGBClassifier(objective='multi:softmax', num_class=len(data['Treatment_Status'].unique()))

# Train the model
xgb_model.fit(xtrain, ytrain)

# Make predictions
ypred_xgb = xgb_model.predict(xtest)

# Evaluate the model
accuracy_xgb = accuracy_score(ytest, ypred_xgb)
print(f'Accuracy: {accuracy_xgb:.2f}')

print('\nClassification Report:')
print(classification_report(ytest, ypred_xgb))

print('\nConfusion Matrix:')
print(confusion_matrix(ytest, ypred_xgb))
```
